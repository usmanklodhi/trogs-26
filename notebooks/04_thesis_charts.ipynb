# 04 · Thesis charts

Class-imbalance chart for the thesis (counts from Table 2, training split).

```python
import matplotlib.pyplot as plt

# Corrected Data extracted EXCLUSIVELY from the Train row of Table 2
characters = ['Alpha', 'Omicron', 'Iota', 'Xi', 'Zeta', 'Psi', 'Upsilon']
counts = [11468, 11030, 10965, 316, 250, 192, 192] 

plt.figure(figsize=(10, 6))
# Using blue for the high-frequency classes, red for the severely underrepresented ones
colors = ['#4C72B0', '#4C72B0', '#4C72B0', '#C44E52', '#C44E52', '#C44E52', '#C44E52']
plt.bar(characters, counts, color=colors)

plt.title('Character Frequency Distribution (Training Set)', fontsize=14)
plt.ylabel('Number of Instances', fontsize=12)
plt.xlabel('Greek Character', fontsize=12)

# Add a reference line to emphasize the massive drop-off
plt.axhline(y=1000, color='gray', linestyle='--', alpha=0.7)

# Add exact numbers on top of the bars for maximum clarity in your thesis
for i, v in enumerate(counts):
    plt.text(i, v + 200, str(v), ha='center', fontsize=10)

plt.tight_layout()
plt.savefig('class_imbalance_chart.png', dpi=300)
plt.show()
```
