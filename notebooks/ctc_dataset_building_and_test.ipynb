```python
import os
import cv2
import random
import numpy as np
import matplotlib.pyplot as plt

# --- CONFIGURATION ---
# Pick ONE image and annotation pair to test the pipeline
IMG_PATH = "data/raw/Images/Agora_I_4985_Rotation1_300dpi.png" # Change to a file you have
ANN_PATH = "data/raw/Annotations/Agora_I_4985_Rotation1_300dpi_letters.txt" # Change to a file you have

print(f"Testing Image: {os.path.basename(IMG_PATH)}")
```

```python
def load_raw_data(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    boxes = []
    transcript_start_idx = -1

    # 1. Extract 8-Point Boxes
    for i, line in enumerate(lines):
        if line.strip().startswith("# Boxes"):
            j = i + 1
            while j < len(lines):
                if lines[j].strip().startswith("# Transcript"):
                    transcript_start_idx = j + 1
                    break
                parts = lines[j].strip().split()
                if len(parts) == 8:
                    try:
                        c = [float(p) for p in parts]
                        poly = np.array([[c[0], c[1]], [c[2], c[3]], [c[4], c[5]], [c[6], c[7]]], dtype=np.float32)
                        boxes.append(poly)
                    except ValueError:
                        pass
                j += 1
            break

    # 2. Extract Raw Transcript Lines
    raw_transcripts = []
    if transcript_start_idx != -1:
        for i in range(transcript_start_idx, len(lines)):
            line_content = lines[i].strip()
            if line_content.startswith("# Lines"):
                break
            if line_content: # Ignore empty lines
                raw_transcripts.append(line_content)

    return boxes, raw_transcripts

boxes, raw_transcripts = load_raw_data(ANN_PATH)
print(f"Total Boxes Found: {len(boxes)}")
print(f"Total Transcript Lines Found: {len(raw_transcripts)}")
```
